```python
import pandas as pd

parsed_foods = []

for food in data['SRLegacyFoods']:
    row = {
        'fdc_id': food.get('fdcId'),
        'description': food.get('description'),
        'category': food.get('foodCategory', {}).get('description')
    }

    for item in food.get('foodNutrients', []):
        nutrient_info = item.get('nutrient') or {}
        name = (nutrient_info.get('name') or '').strip()
        unit = (nutrient_info.get('unitName') or '').strip().lower()
        amount = item.get('amount')

        if amount is None:
            continue

        # Макронутрієнти & Енергія
        if name == 'Energy' and unit == 'kcal':
            row['calories_kcal'] = amount
        elif name == 'Protein':
            row['protein_g'] = amount
        elif name == 'Total lipid (fat)':
            row['fat_g'] = amount
        elif name == 'Carbohydrate, by difference':
            row['carbs_g'] = amount
        elif name == 'Fiber, total dietary':
            row['fiber_g'] = amount
        elif name == 'Total Sugars':
            row['sugars_g'] = amount
        elif name == 'Fatty acids, total saturated':
            row['saturated_fat_g'] = amount
        elif name == 'Fatty acids, total trans':
            row['trans_fat_g'] = amount
        elif name == 'Cholesterol':
            row['cholesterol_mg'] = amount
        elif name == 'Water':
            row['water_g'] = amount
        elif name == 'Ash':
            row['ash_g'] = amount

        # Мінерали
        elif name == 'Sodium, Na':
            row['sodium_mg'] = amount
        elif name == 'Potassium, K':
            row['potassium_mg'] = amount
        elif name == 'Calcium, Ca':
            row['calcium_mg'] = amount
        elif name == 'Iron, Fe':
            row['iron_mg'] = amount
        elif name == 'Magnesium, Mg':
            row['magnesium_mg'] = amount
        elif name == 'Phosphorus, P':
            row['phosphorus_mg'] = amount
        elif name == 'Zinc, Zn':
            row['zinc_mg'] = amount

        # Вітаміни (з гнучким пошуком назв та одиниць)
        elif name == 'Vitamin C, total ascorbic acid':
            row['vitamin_c_mg'] = amount

        # Вітамін A (шукаємо RAE в мікрограмах, або IU якщо немає RAE), конвертуємо одиниці виміру
        elif 'Vitamin A' in name:
            if 'RAE' in name or unit in ['µg', 'ug', 'mcg']:
                row['vitamin_a_mcg'] = amount
            elif unit == 'iu' and 'vitamin_a_mcg' not in row:
                row['vitamin_a_mcg'] = round(amount / 3.33, 1)

        # Вітамін D (шукаємо D2 + D3), конвертуємо одиниці виміру
        elif 'Vitamin D' in name:
            if unit in ['µg', 'ug', 'mcg']:
                row['vitamin_d_mcg'] = amount
            elif unit == 'iu' and 'vitamin_d_mcg' not in row:
                row['vitamin_d_mcg'] = round(amount / 40, 2)

        # Вітамін B-12
        elif 'Vitamin B-12' in name:
            row['vitamin_b12_mcg'] = amount

        elif name == 'Vitamin E (alpha-tocopherol)':
            row['vitamin_e_mg'] = amount
        elif name == 'Vitamin B-6':
            row['vitamin_b6_mg'] = amount
        elif name == 'Caffeine':
            row['caffeine_mg'] = amount

    parsed_foods.append(row)

# Створюємо чистий DataFrame
df = pd.DataFrame(parsed_foods)
```

```python
import numpy as np

# 1. Розмічаємо тип обробки (Raw vs Cooked/Processed)
df_clean = df_step2.copy()

# Регулярні вирази для пошуку ключових слів

raw_pattern = r'\b(?:raw|fresh|dry|dried|unprocessed)\b'
cooked_pattern = r'\b(?:cooked|boiled|roasted|canned|baked|fried|steamed|grilled|toasted|broiled)\b'
# Визначення умов
conditions = [
    # Якщо є маркери сирого/свіжого І немає маркерів термообробки
    df_clean['description'].str.contains(raw_pattern, case=False, na=False) &
    ~df_clean['description'].str.contains(cooked_pattern, case=False, na=False),

    # Якщо є явні маркери термообробки чи консервації
    df_clean['description'].str.contains(cooked_pattern, case=False, na=False)
]

choices = ['Raw / Whole Food', 'Cooked / Processed']

# За замовчуванням (якщо немає жодного слова-маркера) ставимо базову групу
df_clean['processing_type'] = np.select(conditions, choices, default='Raw / Whole Food')

# 2. Функція для створення макро-груп
def assign_macro_group(cat):
    if pd.isna(cat):
        return 'Other'

    cat_lower = str(cat).lower()

    # Рослинна їжа
    if 'vegetable' in cat_lower or 'legume' in cat_lower:
        return 'Vegetables & Legumes'
    elif 'fruit' in cat_lower:
        return 'Fruits'
    elif 'nut' in cat_lower or 'seed' in cat_lower:
        return 'Nuts & Seeds'

    # Тваринна білкова їжа
    elif 'finfish' in cat_lower or 'shellfish' in cat_lower:
        return 'Seafood & Fish'
    elif any(x in cat_lower for x in ['beef', 'pork', 'poultry', 'lamb', 'sausages', 'veal', 'game', 'meat']):
        return 'Meat & Poultry'

    # Інші групи
    elif any(x in cat_lower for x in ['dairy', 'egg']):
        return 'Dairy & Eggs'
    elif any(x in cat_lower for x in ['cereal', 'baked', 'grain']):
        return 'Grains & Bakery'
    elif any(x in cat_lower for x in ['snack', 'sweets', 'beverages', 'fast foods']):
        return 'Snacks, Sweets & Fast Food'
    elif 'fats' in cat_lower or 'oils' in cat_lower:
        return 'Fats & Oils'
    elif any(x in cat_lower for x in ['soups', 'sauces', 'spices']):
        return 'Soups, Sauces & Spices'
    else:
        return 'Other'

# Застосовуємо оновлені макро-групи
df_clean['macro_group'] = df_clean['category'].apply(assign_macro_group)

# Дивимося оновлений розподіл
print("--- РОЗПОДІЛ ЗА МАКРО-ГРУПАМИ ---")
print(df_clean['macro_group'].value_counts())

print("\n--- РОЗПОДІЛ ЗА ТИПОМ ОБРОБКИ ---")
print(df_clean['processing_type'].value_counts())

print("\n--- OTHER ---")
print(df_clean['macro_group'].value_counts().get('Other', 0))
```

```python
import numpy as np

df_clean = df_clean.copy()

# ============================================================
# 1. ЯВНО НЕГОТОВИЙ / СИРИЙ ПРОДУКТ
# ============================================================

raw_pattern = (
    r'\b(?:raw|fresh|unprocessed|uncooked|unheated|'
    r'unprepared|unbaked)\b'
)


# ============================================================
# 2. ЯВНО ПРИГОТОВЛЕНИЙ / ГОТОВИЙ ПРОДУКТ
# ============================================================

cooked_pattern = (
    r'\b(?:cooked|boiled|roasted|baked|fried|steamed|'
    r'grilled|toasted|broiled|microwaved|'
    r'ready[- ]?to[- ]?(?:eat|heat|serve)|'
    r'prepared|commercially prepared|'
    r'prepared[- ]from[- ]recipe|'
    r'canned|pasteurized|'
    r'pickled|preserved)\b'
)


# ============================================================
# 3. ПРОДУКТИ, ЯКІ ЗА НАЗВОЮ Є ГОТОВИМИ
#
# Використовуємо тільки для продуктів, які зазвичай
# продаються/споживаються вже готовими.
# ============================================================

ready_product_pattern = (
    r'\b(?:'

    # Хлібобулочні
    r'bread|'
    r'bagels?|'
    r'rolls?|'
    r'buns?|'
    r'muffins?|'
    r'cookies?|'
    r'crackers?|'
    r'cakes?|'
    r'pastr(?:y|ies)|'
    r'waffles?|'
    r'pancakes?|'
    r'doughnuts?|'
    r'pretzels?|'
    r'tostadas?|'
    r'tortillas?|'
    r'focaccia|'

    # Готові снеки
    r'chips?|'
    r'granola bars?|'
    r'fruit leather|'
    r'beef jerky|'
    r'popcorn|'
    r'rice cakes?|'
    r'trail mix|'

    # Солодощі
    r'cand(?:y|ies)|'
    r'chocolate|'
    r'fudge|'
    r'marshmallows?|'
    r'caramels?|'
    r'jellybeans?|'
    r'gumdrops?|'

    # Десерти
    r'ice cream|'
    r'ice creams|'
    r'sherbet|'
    r'frozen yogurt|'
    r'pudding|'
    r'puddings?|'
    r'flan|'
    r'eclair|'
    r'cream puff|'

    # Готові страви / продукти
    r'pizza|'
    r'pie|'
    r'pies|'
    r'sandwich|'
    r'sandwiches|'

    # Напої / готові продукти
    r'smoothie|'
    r'smoothies|'
    r'juice|'
    r'nectar|'

    r')\b'
)


# ============================================================
# 4. ЯВНО НЕГОТОВІ ПРОДУКТИ
#
# Ці слова мають вищий пріоритет, ніж ready_product_pattern.
# Наприклад:
#
# "refrigerated dough"
# "ready-to-bake"
# "ready-to-fry"
#
# не повинні стати готовим продуктом тільки через
# слово "bread", "tortilla" тощо.
# ============================================================

not_ready_pattern = (
    r'\b(?:'
    r'dough|'
    r'batter|'
    r'raw|'
    r'uncooked|'
    r'unprepared|'
    r'unbaked|'
    r'unheated|'
    r'ready[- ]?to[- ]?(?:bake|fry|cook)|'
    r'refrigerated dough|'
    r'dry mix|'
    r'dry powdered mix|'
    r'dry powder'
    r')\b'
)


# ============================================================
# 5. FROZEN
# ============================================================

frozen_pattern = r'\bfrozen\b'


# Frozen + явно неготовий
frozen_raw_pattern = (
    r'(?=.*\bfrozen\b)'
    r'(?=.*\b(?:raw|uncooked|unprepared|unbaked|unheated)\b)'
)


# Frozen + явно готовий
frozen_cooked_pattern = (
    r'(?=.*\bfrozen\b)'
    r'(?=.*\b(?:cooked|ready[- ]?to[- ]?(?:eat|heat|serve)|'
    r'microwaved|toasted)\b)'
)


# ============================================================
# 6. СТВОРЮЄМО УМОВИ
# ============================================================

raw_condition = df_clean['description'].str.contains(
    raw_pattern,
    case=False,
    na=False,
    regex=True
)

cooked_condition = df_clean['description'].str.contains(
    cooked_pattern,
    case=False,
    na=False,
    regex=True
)

ready_product_condition = df_clean['description'].str.contains(
    ready_product_pattern,
    case=False,
    na=False,
    regex=True
)

not_ready_condition = df_clean['description'].str.contains(
    not_ready_pattern,
    case=False,
    na=False,
    regex=True
)

frozen_raw_condition = df_clean['description'].str.contains(
    frozen_raw_pattern,
    case=False,
    na=False,
    regex=True
)

frozen_cooked_condition = df_clean['description'].str.contains(
    frozen_cooked_pattern,
    case=False,
    na=False,
    regex=True
)


# ============================================================
# 7. КЛАСИФІКАЦІЯ
#
# Порядок дуже важливий!
# ============================================================

conditions = [

    # 1. Frozen + явно неготовий
    frozen_raw_condition,

    # 2. Будь-який явно неготовий продукт
    not_ready_condition,

    # 3. Явно сирий
    raw_condition & ~cooked_condition,

    # 4. Frozen + явно готовий
    frozen_cooked_condition,

    # 5. Явно приготовлений / оброблений
    cooked_condition,

    # 6. Продукти, які за назвою є готовими
    #    але НЕ мають маркера неготовності
    ready_product_condition & ~not_ready_condition
]


choices = [
    'Raw / Whole Food',
    'Raw / Whole Food',
    'Raw / Whole Food',
    'Cooked / Processed',
    'Cooked / Processed',
    'Cooked / Processed'
]


# ============================================================
# 8. ФІНАЛЬНА КЛАСИФІКАЦІЯ
# ============================================================

df_clean['processing_type'] = np.select(
    conditions,
    choices,
    default='Unknown'
)


# ============================================================
# 9. РОЗПОДІЛ
# ============================================================

print("--- РОЗПОДІЛ ЗА ТИПОМ ОБРОБКИ ---")

print(
    df_clean['processing_type']
    .value_counts()
)


print("\n--- ПРОПОРЦІЇ ---")

print(
    df_clean['processing_type']
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)
```

Виявлення аномалій (Outlier Detection) за допомогою IQR
Метод міжквартильного розмаху (Interquartile Range, IQR) дозволяє знайти продукти, де вміст певних нутрієнтів виходить за межі нормального розподілу

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

df = df_clean.copy()

def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    print(f"--- Аномалії для метрики: {column} ---")
    print(f"Знайдено аномальних записів: {len(outliers)} з {len(df)}")
    print(f"Порогові значення: мінімум = {lower_bound:.2f}, максимум = {upper_bound:.2f}\n")
    return outliers

# Перевіримо
for i in columns_check:
    detect_outliers_iqr(df, i)
```

Проведено аналіз аномалій методом IQR. Виявлені верхні відхилення (наприклад, по вітамінах та кофеїну) визнано природною концентрацією речовин у специфічних групах продуктів (фрукти, кава, спеції), а не помилками даних. Нижні межі математично обмежено 0. Подальше вилучення рядків не проводилося задля збереження цілісності категорій

```python
# Розділяємо масиви за створеним раніше processing_type
raw_energy = df[df['processing_type'] == 'Raw / Whole Food']['calories_kcal'].dropna()
cooked_energy = df[df['processing_type'] == 'Cooked / Processed']['calories_kcal'].dropna()

# 1. Перевірка на нормальність розподілу (Shapiro-Wilk)
stat_raw, p_raw = stats.shapiro(raw_energy.sample(min(500, len(raw_energy))))
stat_cooked, p_cooked = stats.shapiro(cooked_energy.sample(min(500, len(cooked_energy))))

# Якщо p-value < 0.05, розподіл не є нормальним — використовуємо непараметричний Mann-Whitney U test
if p_raw < 0.05 or p_cooked < 0.05:
    stat, p_value = stats.mannwhitneyu(raw_energy, cooked_energy)
    test_name = "Mann-Whitney U Test"
else:
    stat, p_value = stats.ttest_ind(raw_energy, cooked_energy)
    test_name = "Independent t-Test"

print(f"--- Результати тестування ({test_name}) ---")
print(f"Статистика: {stat:.4f}, p-value: {p_value:.4e}")

if p_value < 0.05:
    print("Висновок: Нульову гіпотезу відхилено. Кулінарна обробка має статистично значущий вплив на калорійність (p < 0.05).")
else:
    print("Висновок: Немає підстав відхилити нульову гіпотезу. Відмінності не є статистично значущими.")
```

```python
import pandas as pd
import numpy as np
from scipy import stats
from statsmodels.stats.multitest import multipletests

df = df_clean.copy()
# ============================================================
# 1. СПИСОК НУТРІЄНТІВ ДЛЯ ПОРІВНЯННЯ
# ============================================================

#nutrient_cols


# ============================================================
# 2. СПИСОК MACRO GROUP
# ============================================================

macro_groups = df['macro_group'].dropna().unique()


# ============================================================
# 3. ПУСТИЙ СПИСОК ДЛЯ РЕЗУЛЬТАТІВ
# ============================================================

results = []


# ============================================================
# 4. ОСНОВНИЙ ЦИКЛ
# ============================================================

for group in macro_groups:

    # Вибираємо конкретну macro_group
    group_df = df[df['macro_group'] == group]

    # Розділяємо Raw / Cooked
    raw_df = group_df[
        group_df['processing_type'] == 'Raw / Whole Food'
    ]

    cooked_df = group_df[
        group_df['processing_type'] == 'Cooked / Processed'
    ]


    # --------------------------------------------------------
    # Перебираємо всі нутрієнти
    # --------------------------------------------------------

    for nutrient in nutrient_cols:

        raw = raw_df[nutrient].dropna()
        cooked = cooked_df[nutrient].dropna()


        # ----------------------------------------------------
        # Перевірка мінімального розміру вибірки
        # ----------------------------------------------------

        if len(raw) < 5 or len(cooked) < 5:
            continue


        # ----------------------------------------------------
        # Mann-Whitney U test
        # ----------------------------------------------------

        statistic, p_value = stats.mannwhitneyu(
            raw,
            cooked,
            alternative='two-sided'
        )


        # ----------------------------------------------------
        # Медіани
        # ----------------------------------------------------

        raw_median = raw.median()
        cooked_median = cooked.median()


        # ----------------------------------------------------
        # Різниця медіан
        #
        # Позитивне значення:
        # Cooked > Raw
        #
        # Негативне:
        # Raw > Cooked
        # ----------------------------------------------------

        median_difference = cooked_median - raw_median


        # ----------------------------------------------------
        # Effect Size
        #
        # Rank-Biserial Correlation
        #
        # r = 1  -> Cooked повністю вище Raw
        # r = 0  -> практично немає різниці
        # r = -1 -> Raw повністю вище Cooked
        # ----------------------------------------------------

        n_raw = len(raw)
        n_cooked = len(cooked)

        rank_biserial = (
            (2 * statistic) / (n_raw * n_cooked)
        ) - 1


        # ----------------------------------------------------
        # Середні значення — додатково для Power BI
        # ----------------------------------------------------

        raw_mean = raw.mean()
        cooked_mean = cooked.mean()


        # ----------------------------------------------------
        # Зберігаємо результат
        # ----------------------------------------------------

        results.append({
            'macro_group': group,
            'nutrient': nutrient,

            'raw_n': n_raw,
            'cooked_n': n_cooked,

            'raw_median': raw_median,
            'cooked_median': cooked_median,
            'median_difference': median_difference,

            'raw_mean': raw_mean,
            'cooked_mean': cooked_mean,

            'u_statistic': statistic,
            'p_value': p_value,

            'effect_size': rank_biserial
        })


# ============================================================
# 5. СТВОРЮЄМО DATAFRAME
# ============================================================

results_df = pd.DataFrame(results)


# ============================================================
# 6. BENJAMINI-HOCHBERG CORRECTION
# ============================================================

results_df['p_adjusted'] = multipletests(
    results_df['p_value'],
    method='fdr_bh'
)[1]


# ============================================================
# 7. СТАТИСТИЧНА ЗНАЧУЩІСТЬ ПІСЛЯ КОРЕКЦІЇ
# ============================================================

results_df['significant'] = (
    results_df['p_adjusted'] < 0.05
)


# ============================================================
# 8. НАПРЯМОК РІЗНИЦІ
# ============================================================

results_df['direction'] = np.where(
    results_df['median_difference'] > 0,
    'Cooked > Raw',
    np.where(
        results_df['median_difference'] < 0,
        'Raw > Cooked',
        'No difference'
    )
)


# ============================================================
# 9. ІНТЕРПРЕТАЦІЯ EFFECT SIZE
# ============================================================

def effect_interpretation(effect):

    abs_effect = abs(effect)

    if abs_effect < 0.1:
        return 'Negligible'

    elif abs_effect < 0.3:
        return 'Small'

    elif abs_effect < 0.5:
        return 'Medium'

    else:
        return 'Large'


results_df['effect_interpretation'] = (
    results_df['effect_size']
    .apply(effect_interpretation)
)


# ============================================================
# 10. СОРТУВАННЯ
# ============================================================

results_df = results_df.sort_values(
    ['p_adjusted', 'macro_group', 'nutrient']
).reset_index(drop=True)


# ============================================================
# 11. РЕЗУЛЬТАТ
# ============================================================

print("Кількість статистичних порівнянь:", len(results_df))

display(results_df)
#results_df.to_csv(
#    'results_df.csv',
#    index=False,
#    encoding='utf-8',
#)
#print("Файл 'results_df.csv' успішно збережено!")

#files.download('results_df.csv')
```

```python
# Захист від ділення на 0 для низькокалорійних продуктів/води
df = df_clean.copy()
calories_safe = df['calories_kcal'].replace(0, np.nan)

# ==========================================
# 1. Macro-Nutrient Caloric Split (% від калорійності)
# ==========================================
# Розраховуємо, який % калорій дають білки (4 ккал/г), жири (9 ккал/г) та вуглеводи (4 ккал/г)
df['protein_cal_pct'] = ((df['protein_g'] * 4) / calories_safe * 100).fillna(0).clip(0, 100).round(1)
df['fat_cal_pct'] = ((df['fat_g'] * 9) / calories_safe * 100).fillna(0).clip(0, 100).round(1)
df['carb_cal_pct'] = ((df['carbs_g'] * 4) / calories_safe * 100).fillna(0).clip(0, 100).round(1)


# ==========================================
# 2. Nutrient Density Score (Індекс користі на 100 ккал)
# ==========================================
# Оцінюємо скільки корисних речовин (білок, клітковина, залізо, кальцій, вітамін C)
# припадає на кожні 100 ккал продукту. Чим вище бал — тим "чистіший" і корисніший продукт.
df['nutrient_density_score'] = (
    (df['protein_g'] * 2 + df['fiber_g'] * 5 + df['iron_mg'] * 2 + df['calcium_mg'] * 0.1 + df['vitamin_c_mg'] * 0.5) /
    (calories_safe / 100)
).fillna(0).round(2)


# ==========================================
# 3. Dietary & Nutrition Flags (Дієтичні теги)
# ==========================================
# Створюємо бінарні прапорці (True/False) для створення швидких сегментів у Tableau
df['is_high_protein'] = (df['protein_cal_pct'] >= 25) | (df['protein_g'] >= 15)
df['is_keto_friendly'] = (df['carb_cal_pct'] <= 10) & (df['fat_cal_pct'] >= 65)
df['is_low_calorie'] = df['calories_kcal'] <= 100
df['is_high_fiber'] = df['fiber_g'] >= 5

# Перевірка створених метрик
print("=== Створено нові колонки ===")
print(df[['protein_cal_pct', 'fat_cal_pct', 'carb_cal_pct', 'nutrient_density_score', 'is_high_protein']].head())
```
